# Airport Point Mapping

Reference notebook — demonstrates plotting point data with GeoPandas and matplotlib.  
Uses the Natural Earth airports dataset (bundled with GeoPandas).

```python
import geopandas as gpd
import matplotlib.pyplot as plt
```

## Load World Countries (base layer)

```python
world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))
world.head()
```

## Load Airports

```python
airports = gpd.read_file(gpd.datasets.get_path('naturalearth_cities'))
print(airports.crs)
airports.head()
```

## Plot Points Over Countries

```python
fig, ax = plt.subplots(figsize=(14, 7))

world.plot(ax=ax, color='lightgray', edgecolor='white', linewidth=0.5)
airports.plot(ax=ax, color='red', markersize=5, alpha=0.7, label='Cities')

ax.set_title('World Cities — Point Mapping with GeoPandas', fontsize=14)
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.legend()
plt.tight_layout()
plt.show()
```

## Filter to a Region

```python
north_america = world[world['continent'] == 'North America']

# Clip airports to North America bounding box
na_airports = airports.cx[-170:-50, 15:75]

fig, ax = plt.subplots(figsize=(10, 7))
north_america.plot(ax=ax, color='lightyellow', edgecolor='gray', linewidth=0.5)
na_airports.plot(ax=ax, color='steelblue', markersize=8, alpha=0.8)

ax.set_title('North American Cities')
plt.tight_layout()
plt.show()
```

## Inspect Geometry

```python
print('Geometry type:', airports.geom_type.unique())
print('CRS:', airports.crs)
print('Bounding box:', airports.total_bounds)
```

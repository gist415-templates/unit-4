## Functions

```python
def haversine_approx(lat1, lon1, lat2, lon2):
    """Very rough distance in km between two lat/lon points."""
    import math
    R = 6371
    dlat = math.radians(lat2 - lat1)
    dlon = math.radians(lon2 - lon1)
    a = math.sin(dlat/2)**2 + math.cos(math.radians(lat1)) * math.cos(math.radians(lat2)) * math.sin(dlon/2)**2
    return R * 2 * math.atan2(math.sqrt(a), math.sqrt(1 - a))

dist = haversine_approx(32.2319, -110.9501, 33.4484, -112.0740)  # Tucson to Phoenix
print(f'Tucson to Phoenix: ~{dist:.1f} km')
```

## Simple Plot with Matplotlib

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 3, 5]

plt.figure(figsize=(6, 4))
plt.plot(x, y, marker='o')
plt.title('My First Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.tight_layout()
plt.show()
```
